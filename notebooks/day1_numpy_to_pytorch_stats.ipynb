```python
import torch
class Solution():
    def __init__(self, s_data):
        #  Converting data type 
        self.data = s_data.to(dtype=torch.float32)
        print(f"The data has {self.data.shape[1]} columns and {self.data.shape[0]} rows")

    def statistics(self):
        mean = torch.mean(self.data, dim=0)
        std = torch.std(self.data, dim=0)
        #  torch.corrcoef calculates relationships between rows.
        # Since columns are features, transposing with .T 
        correlation = torch.corrcoef(self.data.T)
        return mean, std, correlation

    def outlier(self):
        #  Loop through column indices since tensors don't have text .columns
        num_cols = self.data.shape[1]

        for col in range(num_cols):
            column_data = self.data[:, col] # Selects all rows for this column

            #  Using torch.quantile() global function
            Q1 = torch.quantile(column_data, 0.25)
            Q3 = torch.quantile(column_data, 0.75)
            IQR = Q3 - Q1

            maximum = Q3 + 1.5 * IQR
            minimum = Q1 - 1.5 * IQR

            # PyTorch boolean indexing for outliers
            outliers = column_data[(column_data < minimum) | (column_data > maximum)]

            print(f"Column {col}: {len(outliers)} outliers detected")


import time  
import numpy as np
def run_time():
    rows, columns = 1000000, 10  
    print(f"\n Running Benchmark on {rows:,} rows...")

    # 1. NumPy Benchmark
    strt_up = time.time()
    data_np = np.random.rand(rows, columns)
    mean_np = np.mean(data_np, axis=0)
    correlatin_np = np.corrcoef(data_np.T) 
    end_np = time.time() - strt_up

    # 2. PyTorch Benchmark
    strt_up = time.time()
    data_th = torch.rand(rows, columns)
    mean_th = torch.mean(data_th, dim=0)
    correlatin_th = torch.corrcoef(data_th.T)
    end_torch = time.time() - strt_up

    print(f" NumPy Total Time: {end_np:.4f} seconds")
    print(f" PyTorch CPU Total Time: {end_torch:.4f} seconds")
    print(f" PyTorch is {end_np / end_torch:.2f}x faster than NumPy on CPU.")
s_dat = torch.rand(100, 5) # 100 rows, 5 columns
obj = Solution(s_dat)
obj.statistics()
obj.outlier()

# Run the Speed Test
run_time()

```
