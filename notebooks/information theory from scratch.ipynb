```python
import math

text = "abcdefghijklmnopqrstuvwxyzz"
total_chars = len(text)

# Find all unique characters in our text
unique_chars = set(text)

entropy = 0.0

print("--- Individual Surprise Levels ---")
for char in unique_chars:
    # 1. Count occurrences and find probability
    count = text.count(char)
    prob = count / total_chars
    
    # 2. Calculate Surprise: I(x) = -log(p)
    # Using math.log2 gives us bits (standard in information theory)
    surprise = -math.log2(prob)
    print(f"Character '{char}': Probability = {prob:.3f}, Surprise = {surprise:.3f} bits")
    
    # 3. Add to total Entropy: H(P) = sum( prob * surprise )
    entropy += prob * surprise

print("\n--- Final Results ---")
print(f"Total Entropy of the text: {entropy:.4f} bits per character")

# For cross_entropy and kl_divergence
# --- 1. True Distribution (P) ---
# Let's say we have a simple 3-letter universe: ['a', 'b', 'c']
# This is the ground truth reality
P = {
    'a': 0.70,   # 70% of the time it's 'a'
    'b': 0.20,   # 20% of the time it's 'b'
    'c': 0.10    # 10% of the time it's 'c'
}

# --- 2. Model's Predicted Distribution (Q) ---
# our AI model makes a guess, but it's a bit clumsy and uncertain
Q = {
    'a': 0.40,   # Model thinks 'a' is only 40% likely
    'b': 0.40,   # Model thinks 'b' is 40% likely
    'c': 0.20    # Model thinks 'c' is 20% likely
}

# --- 3. Calculations ---
true_entropy = 0.0
cross_entropy = 0.0
kl_divergence = 0.0

for char in P.keys():
    p_prob = P[char]
    q_prob = Q[char]
    
    # Calculate True Entropy: -p * log2(p)
    true_entropy += -p_prob * math.log2(p_prob)
    
    # Calculate Cross-Entropy: -p * log2(q)
    cross_entropy += -p_prob * math.log2(q_prob)
    
    # Calculate KL Divergence: p * log2(p / q)
    kl_divergence += p_prob * math.log2(p_prob / q_prob)

print("--- Distribution Comparison ---")
print(f"True Entropy H(P)       = {true_entropy:.4f} bits (The minimum possible limit)")
print(f"Cross-Entropy H(P, Q)   = {cross_entropy:.4f} bits (Surprise when using Model Q)")
print(f"KL Divergence D_KL(P||Q)= {kl_divergence:.4f} bits (The penalty/wasted surprise)")

# Double Check the ultimate AI theorem rule: Cross-Entropy = Entropy + KL Divergence
print(f"\nVerification Check: H(P) + D_KL(P||Q) = {true_entropy + kl_divergence:.4f} bits")
```
